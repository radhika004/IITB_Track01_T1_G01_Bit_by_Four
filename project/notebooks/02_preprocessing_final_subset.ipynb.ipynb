1. To train our model, we selected only the important features and created a subset of the dataset.

2. From PSY.csv, we used the columns Question_key, Category, and Response Time, and we added a column for Participant ID.

3. From EEG.csv, we included the brain wave channels (Delta, Theta, Alpha, Beta) recorded at four electrode positions (TP9, AF7, AF8, TP10).

4. For each EEG wave signal while a question was being solved, we calculated the mean, median, standard deviation, minimum, and maximum values, and we also added a Participant ID.

5. From EYE.csv, we used the features ET_GazeLeftx, ET_GazeLefty, ET_GazeRightx, ET_GazeRighty, ET_PupilLeft, ET_PupilRight, ET_DistanceLeft, and ET_DistanceRight.

6. For each eye-tracking feature, we calculated the mean, median, standard deviation, minimum, and maximum values during question solving.

7. From IVT.csv, we used the features Fixation Index, Fixation Duration, Saccade Index, Saccade Amplitude, Gaze Velocity, and Gaze Acceleration, along with the Participant ID.

8. From IVT data, we derived additional features:Average fixation duration, Average saccade amplitude,Total scanpath length (sum of saccade amplitudes)
,Mean gaze velocity, Mean gaze acceleration, Standard deviation of gaze velocity

All four datasets were merged using the Participant ID.
This process was repeated for all 38 participants, and after combining, the final dataset contained a total of 132 features.

```python
import pandas as pd
import glob

wave_cols = ['Delta_TP9', 'Delta_AF7', 'Delta_AF8', 'Delta_TP10',
             'Theta_TP9', 'Theta_AF7', 'Theta_AF8', 'Theta_TP10',
             'Alpha_TP9', 'Alpha_AF7', 'Alpha_AF8', 'Alpha_TP10',
             'Beta_TP9', 'Beta_AF7', 'Beta_AF8', 'Beta_TP10']

files = glob.glob(r"C:\Users\ASUS\OneDrive\Desktop\IITB_internship\STData\STData\*\*_EEG.csv")
all_data = []

for i, file in enumerate(files, start=1):
    df = pd.read_csv(file)

    #  calculate multiple descriptive statistics at once
    df_agg = df.groupby("QuestionKey")[wave_cols].agg(['mean', 'median', 'std', 'min', 'max']).reset_index()

    df_agg.columns = ['_'.join(col).strip() for col in df_agg.columns.values]
    df_agg = df_agg.rename(columns={'QuestionKey_': 'QuestionKey'})

    df_agg["Participant"] = i
    all_data.append(df_agg)

final_df = pd.concat(all_data, ignore_index=True)

print("New DataFrame with aggregated statistical features:")
print(final_df.head())
```

```python
merged = pd.merge(final_df, final_df1, on=["Participant", "QuestionKey"], how="inner")
```

```python
import pandas as pd
import glob

eye_cols = ['ET_GazeLeftx','ET_GazeLefty','ET_GazeRightx','ET_GazeRighty',
            'ET_PupilLeft','ET_PupilRight','ET_DistanceLeft','ET_DistanceRight']

files = glob.glob(r"C:\Users\ASUS\OneDrive\Desktop\IITB_internship\STData\STData\*\*_EYE.csv")

all_eye = []

for i, file in enumerate(files, start=1):
    df_eye = pd.read_csv(file)
    
    df_eye_agg = df_eye.groupby("QuestionKey")[eye_cols].agg(['mean', 'median', 'std', 'min', 'max']).reset_index()
    
    # ('ET_GazeLeftx', 'mean') into 'ET_GazeLeftx_mean'
    df_eye_agg.columns = ['_'.join(col).strip() for col in df_eye_agg.columns.values]
    df_eye_agg = df_eye_agg.rename(columns={'QuestionKey_': 'QuestionKey'})
    df_eye_agg["Participant"] = i # participant number
    all_eye.append(df_eye_agg)

final_eye_df = pd.concat(all_eye, ignore_index=True)

print("New DataFrame with aggregated statistical features for eye-tracking:")
print(final_eye_df.head())
print(f"\nOriginal number of features: {len(eye_cols)}")
print(f"New number of features: {(final_eye_df.shape[1] - 2)}") # Subtract QuestionKey and Participant
```

```python
final = pd.merge(merged,final_eye_df,on=["Participant", "QuestionKey"], how="inner")
```

```python
import pandas as pd
import numpy as np
import glob

# Define the file path pattern
files = glob.glob(r"C:\Users\ASUS\OneDrive\Desktop\IITB_internship\STData\STData\*\*_IVT.csv")

all_ivt_features = []

for i, file in enumerate(files, start=1):
    print(f"Processing file {i}/{len(files)}: {file}")
    df = pd.read_csv(file)
    
    # Data Cleaning
    df.dropna(subset=['QuestionKey'], inplace=True)
    numeric_cols = ['Fixation Index', 'Fixation Duration', 'Saccade Index', 
                    'Saccade Amplitude', 'Gaze Velocity', 'Gaze Acceleration']
    for col in numeric_cols:
        df[col] = pd.to_numeric(df[col], errors='coerce')

    # Count unique fixations and saccades
    fixation_counts = df.groupby('QuestionKey')['Fixation Index'].nunique().to_frame(name='FixationCount')
    saccade_counts = df.groupby('QuestionKey')['Saccade Index'].nunique().to_frame(name='SaccadeCount')
    
    # 2. Calculate statistics for durations, amplitudes, and dynamics
    agg_stats = df.groupby('QuestionKey').agg(
        AvgFixationDuration = ('Fixation Duration', 'mean'),
        AvgSaccadeAmplitude = ('Saccade Amplitude', 'mean'),
        TotalScanpathLength = ('Saccade Amplitude', 'sum'), # Sum of all saccade distances
        MeanGazeVelocity = ('Gaze Velocity', 'mean'),
        MeanGazeAcceleration = ('Gaze Acceleration', 'mean'),
        StdGazeVelocity = ('Gaze Velocity', 'std') # Variability of speed
    )
    
    # 3. Merge all the created features together
    participant_features = fixation_counts.join(saccade_counts, how='outer')
    participant_features = participant_features.join(agg_stats, how='outer')
    
    #participant number
    participant_features['Participant'] = i
    
    all_ivt_features.append(participant_features.reset_index())

final_ivt_df = pd.concat(all_ivt_features, ignore_index=True)

# Fill any potential NaN values that resulted from joins with 0
final_ivt_df.fillna(0, inplace=True)

print("\n--- New DataFrame with Extracted IVT Features ---")
print(final_ivt_df.head())
print(f"\nSuccessfully created {len(final_ivt_df.columns) - 2} new features for {len(final_ivt_df['Participant'].unique())} participants.")
```

```python
final = pd.merge(final,final_ivt_df,on=["Participant", "QuestionKey"], how="inner")
```
