```python
# we will calculate coefficient of variance to compare variance between the waves :
# Coefficient of Variation (CV):  CV makes it scale-independent by dividing the standard deviation by the mean
import numpy as np

# Columns of interest
cols = ['Beta_TP9', 'Beta_AF7', 'Beta_AF8', 'Beta_TP10',
        'Gamma_TP9', 'Gamma_AF7', 'Gamma_AF8', 'Gamma_TP10']

# Dictionary to store results
cv_results = {}

for col in cols:
    values = df_1spl1[col].dropna()       # drop NaN if any
    mean_val = np.mean(values)
    std_val = np.std(values)
    
    # Avoid division by zero if mean is 0
    cv = std_val / abs(mean_val) if mean_val != 0 else np.nan
    
    cv_results[col] = cv

# Convert to a nice DataFrame
import pandas as pd
cv_df = pd.DataFrame.from_dict(cv_results, orient='index', columns=['CV'])
print(cv_df)

# Optional: sort by CV to see most/least stable
print("\nSorted by CV:")
print(cv_df.sort_values(by='CV', ascending=False))
```

```python
import numpy as np
import pandas as pd

cols = ['Beta_TP9', 'Beta_AF7', 'Beta_AF8', 'Beta_TP10',
        'Gamma_TP9', 'Gamma_AF7', 'Gamma_AF8', 'Gamma_TP10']

# Your specific QuestionKeys
question_keys = [
    '1spl1', '1spl2', '1Item1', '1Item2', '1Item3', '1Item4',
    '1Item5', '1Item6', '1Item7', '1Item8', '1Item9', '1Item10',
    '2spl1', '2spl2', '2Item1', '2Item2', '2Item3', '2Item4',
    '2Item5', '2Item6', '2Item7', '2Item8', '2Item9', '2Item10',
    '3spl1', '3spl2', '3Item15', '3Item2', '3Item5', '3Item12',
    '3Item6', '3Item8', '3Item7', '3Item9', '3Item17', '3Item19',
    '3Item11', '3Item4', '3Item13', '3Item10'
]
all_cv_results = {}

for q in question_keys:
    subset = df[df['QuestionKey'] == q]
    cv_results = {}
    
    for col in cols:
        values = subset[col].dropna()
        mean_val = np.mean(values)
        std_val = np.std(values)
        
        cv = std_val / abs(mean_val) if mean_val != 0 else np.nan
        cv_results[col] = cv
    
    all_cv_results[q] = cv_results

cv_df = pd.DataFrame(all_cv_results).T
cv_df_sorted = cv_df.sort_values(by='Gamma_TP9', ascending=False)

print("CV DataFrame shape:", cv_df.shape)
print(cv_df)
```
